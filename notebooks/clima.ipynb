```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_clima = pd.read_csv("../data/clima.csv")
df_clima
```

```python
df_clima.info()
```

GRAFICA GENERAL

```python
plt.figure(figsize=(9,5))

plt.bar(df_clima["variable"], df_clima["valor"])

plt.xlabel("Variable climática")
plt.ylabel("Valor")
plt.title("Variables climáticas y geográficas de la zona de Mancilla")
plt.xticks(rotation=45, ha="right")
plt.grid(axis="y")

plt.show()
```

```python
df_clima_filtrado = df_clima[df_clima["variable"].isin([
    "Temperatura media anual",
    "Precipitación media",
    "Humedad relativa"
])]

df_clima_filtrado
```

GRAFICA CLIMA

```python
plt.figure(figsize=(8,5))

plt.bar(df_clima_filtrado["variable"], df_clima_filtrado["valor"])

plt.xlabel("Variable")
plt.ylabel("Valor")
plt.title("Condiciones climáticas principales")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y")

plt.show()
```

```python
temperatura = df_clima[df_clima["variable"] == "Temperatura media anual"]["valor"].values[0]
precipitacion = df_clima[df_clima["variable"] == "Precipitación media"]["valor"].values[0]
humedad = df_clima[df_clima["variable"] == "Humedad relativa"]["valor"].values[0]

print("Temperatura media anual:", temperatura, "°C")
print("Precipitación media:", precipitacion, "mm/año")
print("Humedad relativa:", humedad, "%")

if temperatura < 14:
    print("La temperatura corresponde a condiciones frías de zona altoandina.")
else:
    print("La temperatura corresponde a condiciones templadas.")

if precipitacion >= 1000:
    print("La precipitación indica una zona con disponibilidad hídrica importante.")
else:
    print("La precipitación indica una zona relativamente seca.")

if humedad >= 75:
    print("La humedad relativa es alta, coherente con condiciones andinas húmedas.")
else:
    print("La humedad relativa es moderada o baja.")
```

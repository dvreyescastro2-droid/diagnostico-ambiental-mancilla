```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_residuos = pd.read_csv("../data/residuos.csv")
df_residuos
```

```python
df_residuos_numerico = df_residuos[df_residuos["valor"].astype(str).str.replace(".", "", regex=False).str.replace(",", "", regex=False).str.isnumeric()].copy()

df_residuos_numerico["valor"] = pd.to_numeric(df_residuos_numerico["valor"])

df_residuos_numerico
```

GRAFICA RESIDUOS

```python
plt.figure(figsize=(8,5))

plt.bar(df_residuos_numerico["variable"], df_residuos_numerico["valor"])

plt.xlabel("Variable")
plt.ylabel("Valor")
plt.title("Generación de residuos sólidos en Facatativá")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y")

plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
agua = pd.read_csv("../data/agua.csv")
clima = pd.read_csv("../data/clima.csv")
residuos = pd.read_csv("../data/residuos.csv")
vegetacion = pd.read_csv("../data/vegetacion.csv")
```

```python
display(agua)
display(clima)
display(residuos)
display(vegetacion)
```

```python
agua.groupby("temporada")[["ph", "temperatura_c", "turbidez_ntu", "conductividad_us_cm", "oxigeno_disuelto_mg_l"]].mean()
```

```python
fig, ax = plt.subplots(figsize=(8,5))

promedios_agua = agua.groupby("temporada")[["ph", "turbidez_ntu", "oxigeno_disuelto_mg_l"]].mean()

promedios_agua.plot(kind="bar", ax=ax)

plt.xlabel("Temporada")
plt.ylabel("Valor promedio")
plt.title("Comparación de variables del agua por temporada")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

```python
ph_promedio = agua["ph"].mean()
turbidez_promedio = agua["turbidez_ntu"].mean()

residuos_mes = 2583.3
residuos_anio = residuos_mes * 12

arbustales = vegetacion[vegetacion["cobertura"] == "Arbustales"]["porcentaje"].values[0]
zonas_intervenidas = vegetacion[vegetacion["cobertura"] == "Zonas industriales o comerciales"]["porcentaje"].values[0]

print("Evaluación integrada del diagnóstico ambiental")

if ph_promedio >= 6.5 and ph_promedio <= 8.5:
    print("El componente agua presenta pH dentro de un rango aceptable.")
else:
    print("El componente agua requiere atención por valores de pH fuera del rango esperado.")

if turbidez_promedio > 5:
    print("La turbidez representa un posible factor de presión sobre la calidad del agua.")
else:
    print("La turbidez promedio no representa una presión elevada.")

if residuos_anio > 30000:
    print("La generación de residuos sólidos representa una presión ambiental municipal importante.")
else:
    print("La generación de residuos sólidos es moderada.")

if arbustales > 50 and zonas_intervenidas > 20:
    print("La zona conserva cobertura vegetal dominante, pero presenta presión por áreas intervenidas.")
else:
    print("La presión sobre la cobertura vegetal es baja o moderada.")
```

# Diagnóstico ambiental del Embalse Mancilla – Facatativá

## Descripción del proyecto

El presente proyecto desarrolla un diagnóstico ambiental básico del Embalse Mancilla y su zona de influencia en el municipio de Facatativá, Cundinamarca, utilizando información secundaria proveniente de artículos científicos, documentos técnicos y reportes institucionales.

El análisis integra componentes relacionados con:

- calidad del agua,
- variables climáticas,
- residuos sólidos,
- cobertura vegetal.

Para el procesamiento y análisis de los datos se utilizaron herramientas computacionales como Python, pandas, matplotlib y Jupyter Notebook. Asimismo, se implementó interpolación lineal como método numérico para estimar valores intermedios entre datos conocidos.

Los datos utilizados corresponden a información secundaria real obtenida de estudios ambientales, documentos institucionales y bases técnicas relacionadas con el Embalse Mancilla, la Ecoreserva Mansilla y el municipio de Facatativá.

## Método numérico utilizado

Se utilizó interpolación lineal para estimar valores intermedios de variables ambientales.

Fórmula:

y = y1 + ((x - x1)(y2 - y1)) / (x2 - x1)

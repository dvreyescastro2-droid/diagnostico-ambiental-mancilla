```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_agua = pd.read_csv("../data/agua.csv")
df_agua
```

```python
df_agua.info()
```

```python
df_agua.describe()
```

GRAFICA DEL PH

```python
plt.figure(figsize=(8,5))
plt.plot(df_agua["distancia_m"], df_agua["ph"], marker="o")
plt.xlabel("Distancia de muestreo (m)")
plt.ylabel("pH")
plt.title("Comportamiento del pH en el Embalse Mancilla")
plt.grid(True)
plt.show()
```

GRAFICA DE TURBIDEZ

```python
plt.figure(figsize=(8,5))
plt.plot(df_agua["distancia_m"], df_agua["turbidez_ntu"], marker="o")
plt.xlabel("Distancia de muestreo (m)")
plt.ylabel("Turbidez (NTU)")
plt.title("Comportamiento de la turbidez en el Embalse Mancilla")
plt.grid(True)
plt.show()
```

GRAFICA DE CONDUCTIVIDAD

```python
plt.figure(figsize=(8,5))
plt.plot(df_agua["distancia_m"], df_agua["conductividad_us_cm"], marker="o")
plt.xlabel("Distancia de muestreo (m)")
plt.ylabel("Conductividad (µS/cm)")
plt.title("Comportamiento de la conductividad en el Embalse Mancilla")
plt.grid(True)
plt.show()
```

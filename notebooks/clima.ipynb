```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_clima = pd.read_csv("../data/clima.csv")
df_clima
```

```python
df_clima.info()
```

```python
df_clima.describe()
```

GRAFICA DE PRECIPITACIÓN

```python
plt.figure(figsize=(8,5))

plt.plot(df_clima["anio"], df_clima["precipitacion_mm"], marker="o")

plt.xlabel("Año")
plt.ylabel("Precipitación (mm)")
plt.title("Precipitación anual en Facatativá")

plt.grid(True)

plt.show()
```

GRAFICA DE TEMPERATURA

```python
plt.figure(figsize=(8,5))

plt.plot(df_clima["anio"], df_clima["temperatura_media_c"], marker="o")

plt.xlabel("Año")
plt.ylabel("Temperatura media (°C)")
plt.title("Temperatura media anual en Facatativá")

plt.grid(True)

plt.show()
```

GRAFICA DE HUMEDAD

```python
plt.figure(figsize=(8,5))

plt.plot(df_clima["anio"], df_clima["humedad_relativa"], marker="o")

plt.xlabel("Año")
plt.ylabel("Humedad relativa (%)")
plt.title("Humedad relativa en Facatativá")

plt.grid(True)

plt.show()
```

```python
x1 = 2022
x2 = 2024
x = 2023

p1 = 1085
p2 = 1150

precipitacion_estimada = p1 + ((x - x1) * (p2 - p1)) / (x2 - x1)

print("La precipitación estimada para 2023 es:", precipitacion_estimada, "mm")
```

## Interpretación del componente clima

Los datos climáticos muestran un incremento progresivo de la precipitación anual y de la humedad relativa en la zona de Facatativá. La temperatura media presenta un comportamiento relativamente estable entre 13 y 14 °C, coherente con las condiciones climáticas de montaña del municipio.

La interpolación lineal permitió estimar la precipitación para el año 2023 a partir de datos conocidos de 2022 y 2024.

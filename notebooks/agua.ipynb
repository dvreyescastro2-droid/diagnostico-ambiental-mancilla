```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_agua = pd.read_csv("../data/agua.csv")
df_agua
```

```python
promedio_temporada = df_agua.groupby("temporada")[["ph", "temperatura_c", "turbidez_ntu", "conductividad_us_cm", "oxigeno_disuelto_mg_l"]].mean()
promedio_temporada
```

GRAFICA DE PH POR TEMPORADA

```python
promedio_temporada["ph"].plot(kind="bar", figsize=(8,5))

plt.xlabel("Temporada")
plt.ylabel("pH promedio")
plt.title("pH promedio por temporada en el Embalse Mancilla")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

GRAFICA DE TURBIDEZ POR TEMPORADA

```python
promedio_temporada["turbidez_ntu"].plot(kind="bar", figsize=(8,5))

plt.xlabel("Temporada")
plt.ylabel("Turbidez promedio (NTU)")
plt.title("Turbidez promedio por temporada")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

GRAFICA OXIGENO DISUELTO POR TEMPORADA

```python
promedio_temporada["oxigeno_disuelto_mg_l"].plot(kind="bar", figsize=(8,5))

plt.xlabel("Temporada")
plt.ylabel("Oxígeno disuelto promedio (mg/L)")
plt.title("Oxígeno disuelto promedio por temporada")
plt.xticks(rotation=0)
plt.grid(axis="y")

plt.show()
```

COMPARACIÓN POR ESTACIÓN

```python
plt.figure(figsize=(8,5))

for estacion in df_agua["estacion"].unique():
    datos = df_agua[df_agua["estacion"] == estacion]
    plt.plot(datos["temporada"], datos["turbidez_ntu"], marker="o", label=estacion)

plt.xlabel("Temporada")
plt.ylabel("Turbidez (NTU)")
plt.title("Turbidez por estación y temporada")
plt.legend()
plt.grid(True)

plt.show()
```

## Interpretación del componente agua

Los valores de pH se encuentran cercanos a la neutralidad, lo cual indica condiciones relativamente estables para el agua del embalse. La turbidez presenta una disminución progresiva entre los puntos de muestreo, lo que puede relacionarse con sedimentación de partículas. La conductividad muestra un aumento leve, indicando presencia de sales o minerales disueltos en el agua.

La interpolación lineal permitió estimar valores intermedios a 150 metros entre los puntos P2 y P3.

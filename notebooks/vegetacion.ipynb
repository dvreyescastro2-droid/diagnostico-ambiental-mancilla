```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_vegetacion = pd.read_csv("../data/vegetacion.csv")
df_vegetacion
```

```python
df_vegetacion.describe()
```

GRAFICA DE BARRAS

```python
plt.figure(figsize=(9,5))

plt.bar(df_vegetacion["cobertura"], df_vegetacion["porcentaje"])

plt.xlabel("Tipo de cobertura")
plt.ylabel("Porcentaje (%)")
plt.title("Coberturas principales de la Ecoreserva Mansilla")
plt.xticks(rotation=35, ha="right")
plt.grid(axis="y")

plt.show()
```

```python
plt.figure(figsize=(7,7))

plt.pie(
    df_vegetacion["porcentaje"],
    labels=df_vegetacion["cobertura"],
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribución porcentual de coberturas en la Ecoreserva Mansilla")
plt.show()
```

GRAFICA DE AREA INTERVENIDA

```python
plt.figure(figsize=(8,5))

plt.plot(df_vegetacion["anio"], df_vegetacion["area_intervenida_ha"], marker="o")

plt.xlabel("Año")
plt.ylabel("Área intervenida (ha)")
plt.title("Cambio del área intervenida")

plt.grid(True)

plt.show()
```

GRAFICA DE COBERTURAS PRINCIPALES

```python
plt.figure(figsize=(8,5))

plt.plot(df_vegetacion["anio"], df_vegetacion["arbustales_pct"], marker="o", label="Arbustales")
plt.plot(df_vegetacion["anio"], df_vegetacion["pastos_limpios_pct"], marker="o", label="Pastos limpios")

plt.xlabel("Año")
plt.ylabel("Porcentaje de cobertura (%)")
plt.title("Coberturas principales en la Ecoreserva Mansilla")

plt.legend()
plt.grid(True)

plt.show()
```

```python
x1 = 2022
x2 = 2024
x = 2023

y1 = 798
y2 = 785

cobertura_estimada = y1 + ((x - x1) * (y2 - y1)) / (x2 - x1)

print("La cobertura vegetal estimada para 2023 es:", cobertura_estimada, "ha")
```

## Interpretación del componente vegetación

La Ecoreserva Mansilla se ubica en la vereda Mancilla del municipio de Facatativá y presenta coberturas asociadas a arbustales y pastos limpios. Estas coberturas son importantes porque protegen el suelo, regulan la humedad, favorecen la biodiversidad y ayudan a conservar el recurso hídrico.

Los datos muestran una disminución progresiva de la cobertura vegetal entre 2020 y 2024, mientras que el área intervenida presenta un aumento. Esto puede interpretarse como una presión sobre el ecosistema, asociada a actividades humanas, cambios de uso del suelo o expansión de áreas transformadas.

La interpolación lineal permitió estimar la cobertura vegetal para el año 2023 a partir de los valores conocidos de 2022 y 2024.

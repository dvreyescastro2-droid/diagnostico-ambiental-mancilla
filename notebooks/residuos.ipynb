```python
import pandas as pd
import matplotlib.pyplot as plt
```

```python
df_residuos = pd.read_csv("../data/residuos.csv")
df_residuos
```

```python
df_residuos_numerico = df_residuos[df_residuos["valor"].astype(str).str.replace(".", "", regex=False).str.replace(",", "", regex=False).str.isnumeric()].copy()

df_residuos_numerico["valor"] = pd.to_numeric(df_residuos_numerico["valor"])

df_residuos_numerico
```

GRAFICA RESIDUOS

```python
plt.figure(figsize=(8,5))

plt.bar(df_residuos_numerico["variable"], df_residuos_numerico["valor"])

plt.xlabel("Variable")
plt.ylabel("Valor")
plt.title("Generación de residuos sólidos en Facatativá")
plt.xticks(rotation=30, ha="right")
plt.grid(axis="y")

plt.show()
```

```python
variables = [
    "Residuos mensuales",
    "Residuos anuales"
]

valores = [
    2583.3,
    30999.6
]

plt.figure(figsize=(8,5))

plt.bar(variables, valores)

plt.xlabel("Tipo de generación")
plt.ylabel("Toneladas")
plt.title("Comparación entre generación mensual y anual de residuos")

plt.grid(axis="y")

plt.show()
```

```python
labels = [
    "Disposición final",
    "Gestión y prestación"
]

sizes = [
    80,
    20
]

plt.figure(figsize=(7,7))

plt.pie(
    sizes,
    labels=labels,
    autopct="%1.1f%%",
    startangle=90
)

plt.title("Distribución general de la gestión de residuos")

plt.show()
```

```python
residuos_mes = 2583.3
residuos_anio = residuos_mes * 12

print("Residuos mensuales:", residuos_mes, "ton/mes")
print("Residuos anuales estimados:", residuos_anio, "ton/año")

if residuos_anio > 30000:
    print("La generación anual de residuos sólidos es alta.")
else:
    print("La generación anual de residuos sólidos es moderada.")

if residuos_mes > 2000:
    print("La generación mensual de residuos representa una presión importante para la gestión municipal.")
else:
    print("La generación mensual de residuos es baja o controlada.")
```

## Interpretación del componente residuos

La generación de residuos sólidos en el municipio de Facatativá representa una presión ambiental importante debido al crecimiento urbano y a la necesidad de disposición final adecuada.

La información del PGIRS indica que los residuos son gestionados principalmente mediante disposición final en el relleno sanitario Nuevo Mondoñedo. Asimismo, la prestación del servicio público de aseo se encuentra a cargo de Trash Global S.A ESP.

Las gráficas permiten observar la magnitud de la generación de residuos y la importancia de fortalecer estrategias de aprovechamiento y gestión integral.

GRAFIA DE RESIDUOS GENERADOS

```python
plt.figure(figsize=(8,5))

plt.plot(df_residuos["anio"], df_residuos["residuos_generados_ton"], marker="o")

plt.xlabel("Año")
plt.ylabel("Residuos generados (toneladas)")
plt.title("Generación anual de residuos sólidos en Facatativá")

plt.grid(True)

plt.show()
```

GRAFICA DE APROVECHAMIENTO

```python
plt.figure(figsize=(8,5))

plt.plot(df_residuos["anio"], df_residuos["aprovechamiento_ton"], marker="o")

plt.xlabel("Año")
plt.ylabel("Residuos aprovechados (toneladas)")
plt.title("Aprovechamiento de residuos sólidos")

plt.grid(True)

plt.show()
```

GRAFICA DE DISPOSICIÓN FINAL

```python
plt.figure(figsize=(8,5))

plt.plot(df_residuos["anio"], df_residuos["disposicion_final_ton"], marker="o")

plt.xlabel("Año")
plt.ylabel("Disposición final (toneladas)")
plt.title("Disposición final de residuos sólidos")

plt.grid(True)

plt.show()
```

## Interpretación del componente residuos

El componente de residuos sólidos permite analizar la generación anual, el aprovechamiento y la disposición final en el municipio de Facatativá. Según la información municipal, Facatativá cuenta con un Plan de Gestión Integral de Residuos Sólidos, PGIRS, como instrumento de planificación ambiental.

Los datos muestran un aumento progresivo en la generación de residuos sólidos entre 2020 y 2024. También se observa un incremento en el aprovechamiento, lo cual indica una mejora en la recuperación de materiales. Sin embargo, la disposición final sigue representando la mayor proporción de los residuos generados.

La interpolación lineal permitió estimar valores intermedios para el año 2023, comparando los datos conocidos de 2022 y 2024.
